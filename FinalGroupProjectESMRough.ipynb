## Introduction

**What is the main theme? Why is it important?**
- 1D tsunami propagation is a way of simulating how a tsunami wave travels through the ocean in one spatial dimension from the ocean towards the coast. This model shows wave propagation in one horizontal direction.
- We do this to understand how a tsunami is moving towards the coast, how its wave height is changing, and how the wave is influenced by the ocean floor as it reaches the coastline.
- We start with a disturbance, for example an earthquake. We then calculate how the wave moves across the ocean using shallow water equations, and observe how the wave changes as ocean depth decreases.
- Modeling 1D tsunami propagation is important because it helps scientists predict, understand, and prepare for tsunamis as they travel across the ocean and impact coastal areas. Instead of modeling the ocean in two or three dimensions, 1D tsunami propagation allows scientists to study tsunamis in one dimension which is a much simpler model.
- Scientists can estimate how quickly a tsunami is traveling, how long it will take to reach the coastline, and to predict arrival times for coastal communities that may need to evacuate.
- Scientists can then improve warning signals. Understanding tsunami waves can tell us which coastal areas are in danger, how much time there is to evacuate, the magnitude of the tsunami, and how high the wave may be.
We can understand how a tsunami changes near the coastline

## Equations
**What are the fundamental physical/chemical/etc. processes?**
- Shallow wave equations can be used to predict tsunami behavior in the open ocean. Shallow Wave equations are approximations to the 3D Navier- Stokes fluid equations.
- Conservation of mass, conservation of momentum, gravity, bathymetry, and nonlinear advection
- Linear shallow water momentum (1D):  $$\frac{\partial u}{\partial t} + g \frac{\partial \eta}{\partial x} = 0$$
- Shallow water wave speed equation: $$c = \sqrt{gH}$$
- Momentum equation:
$$\frac{\partial u}{\partial t} + u \frac{\partial u}{\partial x} = -g \frac{\partial \eta}{\partial x}
$$
- Nonlinear advection:
$$ \frac{\partial^2 u}{\partial t^2} - \alpha^2 \frac{\partial^2 u}{\partial x^2} = 0$$

**Relevant equations and potentially where they come from (i.e. how are they derived)**
- (Kowalik Introduction to Numerical Modeling of Tsunami Waves Textbook)  “Shallow water approximation, probably was applied for the first time by Hansen (1949) to study numerically propagation of tides. It assumes that the horizontal velocities are constant along the vertical direction, also the motion is usually considered to be barotropic i.e., density stratification is neglected.”
